```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('neo4j_query_table_data_q1.csv')
df.columns = df.columns.str.strip()

# Convert Month to datetime
if 'Month' in df.columns:
    df['Month'] = pd.to_datetime(df['Month'], errors='coerce')

sns.set_theme(style="whitegrid")

# Total Volume by Topic and Country
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Country', y='MonthlyVolume', hue='Topic', errorbar=None, estimator=sum)
plt.title('Total Media Volume by Topic and Country')
plt.ylabel('Total Volume')
plt.xlabel('Country')
plt.tight_layout()
plt.savefig('volume_by_topic_country.png')
plt.close()

# Sentiment over time
g = sns.FacetGrid(df, col="Topic", hue="Country", col_wrap=3, height=4, aspect=1.2)
g.map(sns.lineplot, "Month", "MonthlySentiment", marker="o", errorbar=None)
g.add_legend(title="Country")
g.fig.suptitle('Average Media Sentiment Over Time', y=1.05)
for ax in g.axes.flatten():
    ax.axhline(0, color='black', linestyle='--', alpha=0.5)
    ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.savefig('sentiment_over_time.png')
plt.close()

# Volume over time
g2 = sns.FacetGrid(df, col="Topic", hue="Country", col_wrap=3, height=4, aspect=1.2)
g2.map(sns.lineplot, "Month", "MonthlyVolume", marker="o", errorbar=None)
g2.add_legend(title="Country")
g2.fig.suptitle('Media Volume Over Time', y=1.05)
for ax in g2.axes.flatten():
    ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.savefig('volume_over_time.png')
plt.close()

print("Corrected plots saved.")
```
